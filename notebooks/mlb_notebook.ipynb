## Load the data

```python
import pandas as pd
import matplotlib.pyplot as plt

# One row per MLB season, for the last 10 seasons.
seasons = pd.read_csv("/content/mlb_data.csv")

seasons.head()
```

## 1. How often do teams even try to steal a base?

`attempt_rate` is stolen base attempts per team game played that season.

```python
# Plot steal attempt rate across the last 10 seasons
seasons.plot(x="season", y="attempt_rate", kind="line", marker="o", title="Stolen Base Attempt Rate by Season")
plt.ylabel("Attempts per team game")
plt.show()
```
