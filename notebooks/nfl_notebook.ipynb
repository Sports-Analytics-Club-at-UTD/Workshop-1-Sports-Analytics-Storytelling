## 1. How often do teams win when they win the turnover battle?

turnover_margin is positive when a team forced more turnovers than it committed. Let's see how often those teams actually won.

```python
# Teams that won the turnover battle
won_turnover_battle = games[games["turnover_margin"] > 0]

# "win" is already true/false for every row, taking the mean gives the proportion of true values
win_rate_with_to_edge = won_turnover_battle["win"].mean()

print(f"Teams that won the turnover battle won {win_rate_with_to_edge:.1%} of the time")
```
